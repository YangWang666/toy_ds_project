```python
import numpy as np
import pandas as pd
import os
import random
from shutil import copyfile
import pydicom as dicom
import cv2
import glob

# set parameters here
savepath = 'data'
seed = 0
np.random.seed(seed) # Reset the seed so all runs are the same.
random.seed(seed)
MAXVAL = 255  # Range [0 255]

# path to covid-19 dataset from https://github.com/ieee8023/covid-chestxray-dataset
cohen_imgpath = '/content/covid-chestxray-dataset/images'
cohen_csvpath = '/content/covid-chestxray-dataset/metadata.csv'

# path to covid-19 dataset from https://github.com/agchung/Figure1-COVID-chestxray-dataset
fig1_imgpath = '/content/Figure1-COVID-chestxray-dataset/images'
fig1_csvpath = '/content/Figure1-COVID-chestxray-dataset/metadata.csv'

# path to covid-19 dataset from https://github.com/agchung/Actualmed-COVID-chestxray-dataset
actmed_imgpath = '/content/Actualmed-COVID-chestxray-dataset/images'
actmed_csvpath = '/content/Actualmed-COVID-chestxray-dataset/metadata.csv'

# path to https://www.kaggle.com/tawsifurrahman/covid19-radiography-database
sirm_imgpath = '/kaggle/input/covid19-radiography-database/COVID-19_Radiography_Dataset/COVID'
sirm_csvpath = '/kaggle/input/covid19-radiography-database/COVID-19_Radiography_Dataset/COVID.metadata.xlsx'


# path to https://www.kaggle.com/c/rsna-pneumonia-detection-challenge
rsna_datapath = '/content/rsna-data'
# get all the normal from here
rsna_csvname = 'stage_2_detailed_class_info.csv'
# get all the 1s from here since 1 indicate pneumonia
# found that images that aren't pneunmonia and also not normal are classified as 0s
rsna_csvname2 = 'stage_2_train_labels.csv'
rsna_imgpath = 'stage_2_train_images'

# path to ricord covid-19 images created by create_ricord_dataset/create_ricord_dataset.ipynb
# run create_ricord_dataset.ipynb before this notebook
ricord_imgpath = 'create_ricord_dataset/ricord_images'
ricord_txt = 'create_ricord_dataset/ricord_data_set.txt'

# parameters for COVIDx dataset
train = []
test = []
test_count = {'normal': 0, 'pneumonia': 0, 'COVID-19': 0}
train_count = {'normal': 0, 'pneumonia': 0, 'COVID-19': 0}

mapping = dict()
mapping['COVID-19'] = 'COVID-19'
mapping['SARS'] = 'pneumonia'
mapping['MERS'] = 'pneumonia'
mapping['Streptococcus'] = 'pneumonia'
mapping['Klebsiella'] = 'pneumonia'
mapping['Chlamydophila'] = 'pneumonia'
mapping['Legionella'] = 'pneumonia'
mapping['E.Coli'] = 'pneumonia'
mapping['Normal'] = 'normal'
mapping['Lung Opacity'] = 'pneumonia'
mapping['1'] = 'pneumonia'

# train/test split
split = 0.1


# to avoid duplicates
patient_imgpath = {}

# adapted from https://github.com/mlmed/torchxrayvision/blob/master/torchxrayvision/datasets.py#L814
cohen_csv = pd.read_csv(cohen_csvpath, nrows=None)
#idx_pa = csv["view"] == "PA"  # Keep only the PA view
views = ["PA", "AP", "AP Supine", "AP semi erect", "AP erect"]
cohen_idx_keep = cohen_csv.view.isin(views)
cohen_csv = cohen_csv[cohen_idx_keep]

fig1_csv = pd.read_csv(fig1_csvpath, encoding='ISO-8859-1', nrows=None)
actmed_csv = pd.read_csv(actmed_csvpath, nrows=None)

sirm_csv = pd.read_excel(sirm_csvpath)

# get non-COVID19 viral, bacteria, and COVID-19 infections from covid-chestxray-dataset, figure1 and actualmed
# stored as patient id, image filename and label
filename_label = {'normal': [], 'pneumonia': [], 'COVID-19': []}
count = {'normal': 0, 'pneumonia': 0, 'COVID-19': 0}
covid_ds = {'cohen': [], 'fig1': [], 'actmed': [], 'sirm': [], 'ricord': []}

for index, row in cohen_csv.iterrows():
    f = row['finding'].split('/')[-1] # take final finding in hierarchy, for the case of COVID-19, ARDS
    if f == 'COVID-19' and ('eurorad.org' in row['url'] or 'ml-workgroup' in row['url'] or 'sirm.org' in row['url']):
        # skip COVID-19 positive images from eurorad to not duplicate sirm images
        pass
    elif f in mapping: #
        count[mapping[f]] += 1
        entry = [str(row['patientid']), row['filename'], mapping[f], 'cohen']
        filename_label[mapping[f]].append(entry)
        if mapping[f] == 'COVID-19':
            covid_ds['cohen'].append(str(row['patientid']))

for index, row in fig1_csv.iterrows():
    if not str(row['finding']) == 'nan':
        f = row['finding'].split(',')[0] # take the first finding
        if f in mapping: #
            count[mapping[f]] += 1
            if os.path.exists(os.path.join(fig1_imgpath, row['patientid'] + '.jpg')):
                entry = [row['patientid'], row['patientid'] + '.jpg', mapping[f], 'fig1']
            elif os.path.exists(os.path.join(fig1_imgpath, row['patientid'] + '.png')):
                entry = [row['patientid'], row['patientid'] + '.png', mapping[f], 'fig1']
            filename_label[mapping[f]].append(entry)
            if mapping[f] == 'COVID-19':
                covid_ds['fig1'].append(row['patientid'])

for index, row in actmed_csv.iterrows():
    if not str(row['finding']) == 'nan':
        f = row['finding'].split(',')[0]
        if f in mapping:
            count[mapping[f]] += 1
            entry = [row['patientid'], row['imagename'], mapping[f], 'actmed']
            filename_label[mapping[f]].append(entry)
            if mapping[f] == 'COVID-19':
                covid_ds['actmed'].append(row['patientid'])

sirm = set(sirm_csv['URL'])
cohen = set(cohen_csv['url'])
cohen.add('https://github.com/ieee8023/covid-chestxray-dataset')  # Add base URL to remove sirm images from ieee dataset
discard = ['100', '101', '102', '103', '104', '105',
           '110', '111', '112', '113', '122', '123',
           '124', '125', '126', '217']

for idx, row in sirm_csv.iterrows():
    patientid = row['FILE NAME']
    if row['URL'] not in cohen and patientid[patientid.find('(')+1:patientid.find(')')] not in discard:
        count[mapping['COVID-19']] += 1
        imagename = patientid + '.' + row['FORMAT'].lower()
        if not os.path.exists(os.path.join(sirm_imgpath, imagename)):
            imagename = "COVID ({}).png".format(imagename.rsplit(".png")[0].split("COVID ")[1])
        entry = [patientid, imagename, mapping['COVID-19'], 'sirm']
        filename_label[mapping['COVID-19']].append(entry)
        covid_ds['sirm'].append(patientid)

# get ricord file names
with open(ricord_txt) as f:
    ricord_file_names = [line.split()[0] for line in f]

for imagename in ricord_file_names:
    count[mapping['COVID-19']] += 1 # since RICORD data is all COVID-19 postive images
    patientid = imagename.split('-')[3] + '-' + imagename.split('-')[4]

    entry = [patientid, imagename, mapping['COVID-19'], 'ricord']
    filename_label[mapping['COVID-19']].append(entry)

    covid_ds['ricord'].append(patientid)

print('Data distribution from covid datasets:')
print(count)

# Create list of RICORD patients to be added to test, equal to 200 images

# We want to prevent patients present in both train and test
# Get list of patients who have one image
ricord_patients = []
for label in filename_label['COVID-19']:
    if label[3] == 'ricord':
        ricord_patients.append(label[0])

pt_with_one_image = [x for x in ricord_patients if ricord_patients.count(x) == 1] # contrains 176 patients
print(len(pt_with_one_image))

# add covid-chestxray-dataset, figure1 and actualmed into COVIDx dataset
# since these datasets don't have test dataset, split into train/test by patientid
# for covid-chestxray-dataset:
# patient 8 is used as non-COVID19 viral test
# patient 31 is used as bacterial test
# patients 19, 20, 36, 42, 86 are used as COVID-19 viral test
# for figure 1:
# patients 24, 25, 27, 29, 30, 32, 33, 36, 37, 38

ds_imgpath = {'cohen': cohen_imgpath, 'fig1': fig1_imgpath, 'actmed': actmed_imgpath, 'sirm': sirm_imgpath, 'ricord':ricord_imgpath}

for key in filename_label.keys():
    arr = np.array(filename_label[key])
    if arr.size == 0:
        continue
    # split by patients
    # num_diff_patients = len(np.unique(arr[:,0]))
    # num_test = max(1, round(split*num_diff_patients))
    # select num_test number of random patients
    # random.sample(list(arr[:,0]), num_test)
    if key == 'pneumonia':
        test_patients = ['8', '31']
    elif key == 'COVID-19':
        test_patients = ['19', '20', '36', '42', '86',
                         '94', '97', '117', '132',
                         '138', '144', '150', '163', '169', '174', '175', '179', '190', '191'
                         'COVID-00024', 'COVID-00025', 'COVID-00026', 'COVID-00027', 'COVID-00029',
                         'COVID-00030', 'COVID-00032', 'COVID-00033', 'COVID-00035', 'COVID-00036',
                         'COVID-00037', 'COVID-00038',
                         'ANON24', 'ANON45', 'ANON126', 'ANON106', 'ANON67',
                         'ANON153', 'ANON135', 'ANON44', 'ANON29', 'ANON201',
                         'ANON191', 'ANON234', 'ANON110', 'ANON112', 'ANON73',
                         'ANON220', 'ANON189', 'ANON30', 'ANON53', 'ANON46',
                         'ANON218', 'ANON240', 'ANON100', 'ANON237', 'ANON158',
                         'ANON174', 'ANON19', 'ANON195',
                         'COVID 119', 'COVID 87', 'COVID 70', 'COVID 94',
                         'COVID 215', 'COVID 77', 'COVID 213', 'COVID 81',
                         'COVID 216', 'COVID 72', 'COVID 106', 'COVID 131',
                         'COVID 107', 'COVID 116', 'COVID 95', 'COVID 214',
                         'COVID 129']
#         Add 178 RICORD patients to COVID-19, equal to 200 images
        test_patients.extend(pt_with_one_image)
        test_patients.extend(['419639-000025', '419639-001464'])
    else:
        test_patients = []
    print('Key: ', key)
    print('Test patients: ', test_patients)
    # go through all the patients
    for patient in arr:
        if patient[0] not in patient_imgpath:
            patient_imgpath[patient[0]] = [patient[1]]
        else:
            if patient[1] not in patient_imgpath[patient[0]]:
                patient_imgpath[patient[0]].append(patient[1])
            else:
                continue  # skip since image has already been written
        if patient[0] in test_patients:
            if patient[3] == 'sirm':
                image = cv2.imread(os.path.join(ds_imgpath[patient[3]], patient[1]))
                gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
                patient[1] = patient[1].replace(' ', '')
                cv2.imwrite(os.path.join(savepath, 'test', patient[1]), gray)
            else:
                copyfile(os.path.join(ds_imgpath[patient[3]], patient[1]), os.path.join(savepath, 'test', patient[1]))
            test.append(patient)
            test_count[patient[2]] += 1
        else:
            if patient[3] == 'sirm':
                image = cv2.imread(os.path.join(ds_imgpath[patient[3]], patient[1]))
                gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
                patient[1] = patient[1].replace(' ', '')
                cv2.imwrite(os.path.join(savepath, 'train', patient[1]), gray)
            else:
                copyfile(os.path.join(ds_imgpath[patient[3]], patient[1]), os.path.join(savepath, 'train', patient[1]))
            train.append(patient)
            train_count[patient[2]] += 1

print('test count: ', test_count)
print('train count: ', train_count)

# add normal and rest of pneumonia cases from https://www.kaggle.com/c/rsna-pneumonia-detection-challenge
csv_normal = pd.read_csv(os.path.join(rsna_datapath, rsna_csvname), nrows=None)
csv_pneu = pd.read_csv(os.path.join(rsna_datapath, rsna_csvname2), nrows=None)
patients = {'normal': [], 'pneumonia': []}

for index, row in csv_normal.iterrows():
    if row['class'] == 'Normal':
        patients['normal'].append(row['patientId'])

for index, row in csv_pneu.iterrows():
    if int(row['Target']) == 1:
        patients['pneumonia'].append(row['patientId'])

for key in patients.keys():
    arr = np.array(patients[key])
    if arr.size == 0:
        continue
    # split by patients
    # num_diff_patients = len(np.unique(arr))
    # num_test = max(1, round(split*num_diff_patients))
    test_patients = np.load('rsna_test_patients_{}.npy'.format(key)) # random.sample(list(arr), num_test), download the .npy files from the repo.
    # np.save('rsna_test_patients_{}.npy'.format(key), np.array(test_patients))
    for patient in arr:
        if patient not in patient_imgpath:
            patient_imgpath[patient] = [patient]
        else:
            continue  # skip since image has already been written

        ds = dicom.dcmread(os.path.join(rsna_datapath, rsna_imgpath, patient + '.dcm'))
        pixel_array_numpy = ds.pixel_array
        imgname = patient + '.png'
        if patient in test_patients:
            cv2.imwrite(os.path.join(savepath, 'test', imgname), pixel_array_numpy)
            test.append([patient, imgname, key, 'rsna'])
            test_count[key] += 1
        else:
            cv2.imwrite(os.path.join(savepath, 'train', imgname), pixel_array_numpy)
            train.append([patient, imgname, key, 'rsna'])
            train_count[key] += 1

print('test count: ', test_count)
print('train count: ', train_count)

# final stats
print('Final stats')
print('Train count: ', train_count)
print('Test count: ', test_count)
print('Total length of train: ', len(train))
print('Total length of test: ', len(test))

# export to train and test csv
# format as patientid, filename, label, separated by a space
train_file = open("train_split.txt",'w')
for sample in train:
    if len(sample) == 4:
        info = str(sample[0]) + ' ' + sample[1] + ' ' + sample[2] + ' ' + sample[3] + '\n'
    else:
        info = str(sample[0]) + ' ' + sample[1] + ' ' + sample[2] + '\n'
    train_file.write(info)

train_file.close()

test_file = open("test_split.txt", 'w')
for sample in test:
    if len(sample) == 4:
        info = str(sample[0]) + ' ' + sample[1] + ' ' + sample[2] + ' ' + sample[3] + '\n'
    else:
        info = str(sample[0]) + ' ' + sample[1] + ' ' + sample[2] + '\n'
    test_file.write(info)

test_file.close()
```

```python
import os, glob, re, random, shutil, warnings
import numpy as np
import pandas as pd
import cv2
import pydicom as dicom

warnings.filterwarnings("ignore")
seed = 0
random.seed(seed); np.random.seed(seed)

# 输出根目录
savepath = 'data'
os.makedirs(os.path.join(savepath, 'train'), exist_ok=True)
os.makedirs(os.path.join(savepath, 'test'),  exist_ok=True)

MAXVAL = 255  # Range [0..255]

# ===== 2) 数据源路径（按你提供的）=====
# ieee8023 / Cohen
cohen_imgpath = '/content/covid-chestxray-dataset/images'
cohen_csvpath = '/content/covid-chestxray-dataset/metadata.csv'

# Figure1
fig1_imgpath = '/content/Figure1-COVID-chestxray-dataset/images'
fig1_csvpath = '/content/Figure1-COVID-chestxray-dataset/metadata.csv'

# ActualMed
actmed_imgpath = '/content/Actualmed-COVID-chestxray-dataset/images'
actmed_csvpath = '/content/Actualmed-COVID-chestxray-dataset/metadata.csv'

# SIRM / COVID-19 Radiography Database（注意：Colab 通常不在 /kaggle 路径）
sirm_imgpath = '/content/COVID-19_Radiography_Dataset/COVID'
sirm_csvpath = '/content/COVID-19_Radiography_Dataset/COVID.metadata.xlsx'

# RSNA Pneumonia
rsna_datapath = '/content/rsna-data'
rsna_csvname  = '/content/rsna-data/stage_2_detailed_class_info.csv'
rsna_csvname2 = '/content/rsna-data/stage_2_train_labels.csv'
rsna_imgpath  = '/content/rsna-data/stage_2_train_images'

# （可选）RICORD
ricord_imgpath = 'create_ricord_dataset/ricord_images'
ricord_txt     = 'create_ricord_dataset/ricord_data_set.txt'

# ===== 3) 小工具函数 =====
def exists_dir(path, tag):
    if not os.path.isdir(path):
        print(f"\033[33m[WARN]\033[0m {tag} 目录不存在：{path}，此源将被跳过。")
        return False
    return True

def exists_file(path, tag):
    if not os.path.isfile(path):
        print(f"\033[33m[WARN]\033[0m {tag} 文件不存在：{path}，此源将被跳过。")
        return False
    return True

def resolve_sirm_imagename(row, sirm_dir):
    """
    更稳健地从 SIRM 的元数据里找到真实文件名。返回文件名或 None。
    """
    pid = str(row['FILE NAME']).strip()             # e.g. "COVID-19 (123)" or "COVID 119"
    fmt = str(row.get('FORMAT', 'png')).lower()     # 通常是 png，但别完全相信表格

    # 1) 直猜 pid.format
    cand = os.path.join(sirm_dir, f"{pid}.{fmt}")
    if os.path.exists(cand):
        return f"{pid}.{fmt}"

    # 2) 不信后缀，任意后缀（大小写）都搜
    cands = glob.glob(os.path.join(sirm_dir, pid + ".*"))
    if cands:
        return os.path.basename(cands[0])

    # 3) 去空格再搜
    pid_nospace = pid.replace(" ", "")
    cands = glob.glob(os.path.join(sirm_dir, pid_nospace + ".*"))
    if cands:
        return os.path.basename(cands[0])

    # 4) COVID 与 COVID-19 互换，带不带空格的都试
    variants = set([
        pid.replace("COVID-19", "COVID"),
        pid.replace("COVID", "COVID-19"),
        pid_nospace.replace("COVID-19", "COVID"),
        pid_nospace.replace("COVID", "COVID-19"),
    ])
    for v in variants:
        cands = glob.glob(os.path.join(sirm_dir, v + ".*"))
        if cands:
            return os.path.basename(cands[0])

    # 5) 抓括号里的编号，拼常见模板
    m = re.search(r"\((\d+)\)", pid)
    if m:
        n = m.group(1)
        for base in [f"COVID ({n})", f"COVID-19 ({n})", f"COVID{n}", f"COVID-19{n}"]:
            cands = glob.glob(os.path.join(sirm_dir, base + ".*"))
            if cands:
                return os.path.basename(cands[0])

    return None

def save_gray_png(src_path, dst_path):
    img = cv2.imread(src_path, cv2.IMREAD_GRAYSCALE)
    if img is None:
        print(f"\033[33m[WARN]\033[0m 读取失败，跳过：{src_path}")
        return False
    cv2.imwrite(dst_path, img)
    return True

def write_split_file(fname, samples):
    with open(fname, 'w') as f:
        for sample in samples:
            # 允许四元组: [patientid, filename, label, source]
            if len(sample) >= 4:
                info = f"{sample[0]} {sample[1]} {sample[2]} {sample[3]}\n"
            else:
                info = f"{sample[0]} {sample[1]} {sample[2]}\n"
            f.write(info)

# ===== 4) 初始化容器与映射 =====
train, test = [], []
test_count  = {'normal': 0, 'pneumonia': 0, 'COVID-19': 0}
train_count = {'normal': 0, 'pneumonia': 0, 'COVID-19': 0}

mapping = {
    'COVID-19': 'COVID-19',
    'SARS': 'pneumonia',
    'MERS': 'pneumonia',
    'Streptococcus': 'pneumonia',
    'Klebsiella': 'pneumonia',
    'Chlamydophila': 'pneumonia',
    'Legionella': 'pneumonia',
    'E.Coli': 'pneumonia',
    'Normal': 'normal',
    'Lung Opacity': 'pneumonia',
    '1': 'pneumonia'
}

split = 0.1
patient_imgpath = {}
filename_label = {'normal': [], 'pneumonia': [], 'COVID-19': []}
count = {'normal': 0, 'pneumonia': 0, 'COVID-19': 0}
covid_ds = {'cohen': [], 'fig1': [], 'actmed': [], 'sirm': [], 'ricord': []}

# ===== 5) 读 cohen / figure1 / actualmed / sirm / ricord =====
# 5.1 Cohen
if exists_dir(cohen_imgpath, "cohen_imgpath") and exists_file(cohen_csvpath, "cohen_csv"):
    cohen_csv = pd.read_csv(cohen_csvpath)
    views = ["PA", "AP", "AP Supine", "AP semi erect", "AP erect"]
    cohen_csv = cohen_csv[cohen_csv.view.isin(views)]
    for _, row in cohen_csv.iterrows():
        f = str(row.get('finding', '')).split('/')[-1]
        url = str(row.get('url', ''))
        if f == 'COVID-19' and ('eurorad.org' in url or 'ml-workgroup' in url or 'sirm.org' in url):
            # 跳过以免和 SIRM 重复
            continue
        if f in mapping:
            count[mapping[f]] += 1
            entry = [str(row['patientid']), row['filename'], mapping[f], 'cohen']
            filename_label[mapping[f]].append(entry)
            if mapping[f] == 'COVID-19':
                covid_ds['cohen'].append(str(row['patientid']))
else:
    print("\033[33m[WARN]\033[0m 跳过 Cohen 源。")

# 5.2 Figure1
if exists_dir(fig1_imgpath, "fig1_imgpath") and exists_file(fig1_csvpath, "fig1_csv"):
    fig1_csv = pd.read_csv(fig1_csvpath, encoding='ISO-8859-1')
    for _, row in fig1_csv.iterrows():
        finding = str(row.get('finding', ''))
        if finding and finding != 'nan':
            f = finding.split(',')[0]
            if f in mapping:
                count[mapping[f]] += 1
                pid = str(row['patientid'])
                if os.path.exists(os.path.join(fig1_imgpath, pid + '.jpg')):
                    entry = [pid, pid + '.jpg', mapping[f], 'fig1']
                elif os.path.exists(os.path.join(fig1_imgpath, pid + '.png')):
                    entry = [pid, pid + '.png', mapping[f], 'fig1']
                else:
                    continue
                filename_label[mapping[f]].append(entry)
                if mapping[f] == 'COVID-19':
                    covid_ds['fig1'].append(pid)
else:
    print("\033[33m[WARN]\033[0m 跳过 Figure1 源。")

# 5.3 ActualMed
if exists_dir(actmed_imgpath, "actmed_imgpath") and exists_file(actmed_csvpath, "actmed_csv"):
    actmed_csv = pd.read_csv(actmed_csvpath)
    for _, row in actmed_csv.iterrows():
        finding = str(row.get('finding', ''))
        if finding and finding != 'nan':
            f = finding.split(',')[0]
            if f in mapping:
                count[mapping[f]] += 1
                pid = str(row['patientid'])
                entry = [pid, row['imagename'], mapping[f], 'actmed']
                filename_label[mapping[f]].append(entry)
                if mapping[f] == 'COVID-19':
                    covid_ds['actmed'].append(pid)
else:
    print("\033[33m[WARN]\033[0m 跳过 ActualMed 源。")

# 5.4 SIRM
have_sirm = exists_dir(sirm_imgpath, "sirm_imgpath") and exists_file(sirm_csvpath, "sirm_csv")
cohen_urls = set()
if exists_file(cohen_csvpath, "cohen_csv_for_urls"):
    tmp = pd.read_csv(cohen_csvpath)
    cohen_urls = set(tmp.get('url', pd.Series([], dtype=str)).astype(str).tolist())
cohen_urls.add('https://github.com/ieee8023/covid-chestxray-dataset')

if have_sirm:
    sirm_csv = pd.read_excel(sirm_csvpath)
    discard = set(['100','101','102','103','104','105','110','111','112','113','122','123','124','125','126','217'])
    for _, row in sirm_csv.iterrows():
        patientid = str(row['FILE NAME']).strip()
        url = str(row.get('URL', ''))
        # 去重：不和 cohen 重叠；丢弃列表里的一些问题样本
        inner = patientid[patientid.find('(')+1:patientid.find(')')] if '(' in patientid and ')' in patientid else ''
        if (url not in cohen_urls) and (inner not in discard):
            imagename = resolve_sirm_imagename(row, sirm_imgpath)
            if imagename is None:
                print(f"\033[33m[WARN]\033[0m [SIRM] 找不到文件，跳过：{patientid}")
                continue
            count['COVID-19'] += 1
            entry = [patientid, imagename, 'COVID-19', 'sirm']
            filename_label['COVID-19'].append(entry)
            covid_ds['sirm'].append(patientid)
else:
    print("\033[33m[WARN]\033[0m 跳过 SIRM 源。")

# 5.5 RICORD（可选）
if exists_dir(ricord_imgpath, "ricord_imgpath") and exists_file(ricord_txt, "ricord_txt"):
    with open(ricord_txt) as f:
        ricord_file_names = [line.split()[0] for line in f]
    for imagename in ricord_file_names:
        # RICORD 默认都是 COVID 阳性
        count['COVID-19'] += 1
        patientid = imagename.split('-')[3] + '-' + imagename.split('-')[4]
        entry = [patientid, imagename, 'COVID-19', 'ricord']
        filename_label['COVID-19'].append(entry)
        covid_ds['ricord'].append(patientid)
else:
    print("\033[33m[WARN]\033[0m 跳过 RICORD 源。")

print("Data distribution from covid datasets:", count)

# ===== 6) 按论文设定做 test 列表（修复你缺逗号的问题）=====
# pneumonia 测试病人
pneu_test_patients = ['8', '31']

# COVID-19 测试病人（含多源、再加 RICORD 单图病人若存在）
covid_fixed = [
    '19','20','36','42','86','94','97','117','132','138','144','150','163','169','174','175','179','190','191',
    'COVID-00024','COVID-00025','COVID-00026','COVID-00027','COVID-00029','COVID-00030','COVID-00032','COVID-00033','COVID-00035','COVID-00036','COVID-00037','COVID-00038',
    'ANON24','ANON45','ANON126','ANON106','ANON67','ANON153','ANON135','ANON44','ANON29','ANON201','ANON191','ANON234','ANON110','ANON112','ANON73','ANON220','ANON189','ANON30','ANON53','ANON46','ANON218','ANON240','ANON100','ANON237','ANON158','ANON174','ANON19','ANON195',
    'COVID 119','COVID 87','COVID 70','COVID 94','COVID 215','COVID 77','COVID 213','COVID 81','COVID 216','COVID 72','COVID 106','COVID 131','COVID 107','COVID 116','COVID 95','COVID 214','COVID 129'
]

# RICORD: 只把“恰好一张图”的病人分到 test，补齐约 200 张（如果有 RICORD）
ricord_patients = [lab[0] for lab in filename_label['COVID-19'] if len(lab) >= 4 and lab[3] == 'ricord']
pt_with_one_image = [x for x in ricord_patients if ricord_patients.count(x) == 1]
covid_test_patients = list(covid_fixed) + pt_with_one_image + ['419639-000025', '419639-001464']

# ===== 7) 写入 train/test（cohen/fig1/actmed/sirm/ricord）=====
ds_imgpath = {'cohen': cohen_imgpath, 'fig1': fig1_imgpath, 'actmed': actmed_imgpath, 'sirm': sirm_imgpath, 'ricord': ricord_imgpath}

for key in filename_label.keys():
    arr = np.array(filename_label[key], dtype=object)
    if arr.size == 0:
        continue
    if key == 'pneumonia':
        test_patients = set(pneu_test_patients)
    elif key == 'COVID-19':
        test_patients = set(covid_test_patients)
    else:
        test_patients = set()

    for patient in arr:
        pid, imgname, label, source = patient
        # 去重控制
        if pid not in patient_imgpath:
            patient_imgpath[pid] = [imgname]
        else:
            if imgname in patient_imgpath[pid]:
                continue
            patient_imgpath[pid].append(imgname)

        src = os.path.join(ds_imgpath[source], imgname)
        # SIRM 有空格名，保存时去空格以免后续炸裂
        safe_imgname = imgname.replace(' ', '')
        if pid in test_patients:
            if source == 'sirm':
                ok = save_gray_png(src, os.path.join(savepath, 'test', safe_imgname))
                if not ok: continue
            else:
                if not os.path.exists(src):
                    print(f"\033[33m[WARN]\033[0m 缺失文件，跳过：{src}")
                    continue
                shutil.copyfile(src, os.path.join(savepath, 'test', safe_imgname))
            test.append([pid, safe_imgname, label, source])
            test_count[label] += 1
        else:
            if source == 'sirm':
                ok = save_gray_png(src, os.path.join(savepath, 'train', safe_imgname))
                if not ok: continue
            else:
                if not os.path.exists(src):
                    print(f"\033[33m[WARN]\033[0m 缺失文件，跳过：{src}")
                    continue
                shutil.copyfile(src, os.path.join(savepath, 'train', safe_imgname))
            train.append([pid, safe_imgname, label, source])
            train_count[label] += 1

print('test count:', test_count)
print('train count:', train_count)

# ===== 8) 追加 RSNA（normal 与 pneumonia 的大头）=====
rsna_dir = os.path.join(rsna_datapath, rsna_imgpath)
have_rsna = exists_dir(rsna_dir, "rsna_images") and \
            exists_file(os.path.join(rsna_datapath, rsna_csvname), "rsna_csv1") and \
            exists_file(os.path.join(rsna_datapath, rsna_csvname2), "rsna_csv2")

if have_rsna:
    csv_normal = pd.read_csv(os.path.join(rsna_datapath, rsna_csvname))
    csv_pneu   = pd.read_csv(os.path.join(rsna_datapath, rsna_csvname2))
    patients = {'normal': [], 'pneumonia': []}

    for _, row in csv_normal.iterrows():
        if str(row['class']) == 'Normal':
            patients['normal'].append(str(row['patientId']))
    for _, row in csv_pneu.iterrows():
        try:
            if int(row['Target']) == 1:
                patients['pneumonia'].append(str(row['patientId']))
        except:
            pass

    for key in patients.keys():
        arr = np.array(patients[key], dtype=object)
        if arr.size == 0:
            continue

        # 优先用仓库自带的 npy；没有就按种子 split 自动生成并保存
        npy_path = f'rsna_test_patients_{key}.npy'
        if os.path.exists(npy_path):
            test_patients = set(np.load(npy_path, allow_pickle=True).tolist())
        else:
            uniq = sorted(list(set(arr)))
            k = max(1, int(len(uniq) * split))
            random.shuffle(uniq)
            test_patients = set(uniq[:k])
            np.save(npy_path, np.array(list(test_patients)))
            print(f"\033[36m[INFO]\033[0m 生成 {npy_path}，大小：{len(test_patients)}")

        for pid in arr:
            if pid in patient_imgpath:
                # 避免和前面数据重复
                continue
            dcm_path = os.path.join(rsna_dir, pid + '.dcm')
            if not os.path.exists(dcm_path):
                continue

            try:
                ds = dicom.dcmread(dcm_path)
                img = ds.pixel_array
            except Exception as e:
                print(f"\033[33m[WARN]\033[0m 读 DICOM 失败，跳过 {pid}: {e}")
                continue

            imgname = pid + '.png'
            if pid in test_patients:
                cv2.imwrite(os.path.join(savepath, 'test', imgname), img)
                test.append([pid, imgname, key, 'rsna'])
                test_count[key] += 1
            else:
                cv2.imwrite(os.path.join(savepath, 'train', imgname), img)
                train.append([pid, imgname, key, 'rsna'])
                train_count[key] += 1

    print('test count (after RSNA):', test_count)
    print('train count (after RSNA):', train_count)
else:
    print("\033[33m[WARN]\033[0m 跳过 RSNA 源。")

# ===== 9) 最终统计与导出 =====
print('Final stats')
print('Train count:', train_count)
print('Test count :', test_count)
print('Total length of train:', len(train))
print('Total length of test :', len(test))

write_split_file("train_split.txt", train)
write_split_file("test_split.txt",  test)

print("\033[32m[DONE]\033[0m 已生成：train_split.txt, test_split.txt；图片在 data/train 与 data/test。")
```

```python
# =========================================
# COVIDNet-CXR-3 backbone -> 3-class head  (clean graph-safe version)
# 前置：
# 1) data/train, data/test, train_split.txt, test_split.txt 已生成
# 2) 预训练 CXR-3 ckpt 在 models/COVIDNet-CXR-3/ 下，含 .meta/.index/.data
# =========================================
import os, cv2, numpy as np
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()

# -------- 配置 --------
WEIGHT_DIR  = "models/COVIDNet-CXR-3"
TRAIN_SPLIT = "train_split.txt"
VAL_SPLIT   = "test_split.txt"

BATCH_SIZE  = 16
EPOCHS      = 3
LR          = 1e-4
FREEZE_BACKBONE = True

IMG_MEAN = 0.0
IMG_STD  = 1.0

# 你之前日志里看到的 2D 特征层
FEATURE_TENSOR_OVERRIDE = "flatten_last/Reshape:0"
# ---------------------

# ---- 小工具 ----
def to_int(x, default=256):
    try: return int(x)
    except Exception:
        v = getattr(x, "value", None)
        return int(v) if v is not None else default

def load_split(path, subdir):
    assert os.path.exists(path), f"找不到 {path}"
    samples = []
    with open(path, "r") as f:
        for line in f:
            p = line.strip().split()
            if len(p) < 3: continue
            _, fname, label = p[:3]
            full = os.path.join("data", subdir, fname)
            if os.path.exists(full):
                samples.append((full, {"normal":0,"pneumonia":1,"COVID-19":2}[label]))
    return samples

def batch_iter(lst, bs, shuffle=True):
    idx = np.arange(len(lst))
    if shuffle: np.random.shuffle(idx)
    for i in range(0, len(lst), bs):
        sel = idx[i:i+bs]
        yield [lst[j][0] for j in sel], np.array([lst[j][1] for j in sel], np.int64)

def load_images(paths, target_h, target_w, channels):
    th, tw, ch = to_int(target_h), to_int(target_w), to_int(channels, 1)
    arr = np.zeros((len(paths), th, tw, ch), dtype=np.float32)
    for i, p in enumerate(paths):
        im = cv2.imread(p, cv2.IMREAD_GRAYSCALE)
        if im is None:
            im = np.zeros((th, tw), np.uint8)
        else:
            im = cv2.resize(im, (tw, th), interpolation=cv2.INTER_AREA)
        im = (im/255.0 - IMG_MEAN) / max(IMG_STD, 1e-6)
        if ch == 1:
            arr[i, :, :, 0] = im
        else:
            arr[i] = np.stack([im]*ch, axis=-1)
    return arr

def list_flatten_like_tensors(g):
    cands = []
    for op in g.get_operations():
        n = op.name.lower()
        if ("flatten" in n) or ("reshape" in n):
            for t in op.values():
                if t.shape.ndims == 2 and t.name.endswith(":0"):
                    cands.append(t.name)
    return cands
# ---------------------

# 读分割
train_list = load_split(TRAIN_SPLIT, "train")
val_list   = load_split(VAL_SPLIT,   "test")
print("[INFO] train size:", len(train_list), "val size:", len(val_list))

# ckpt 基名
assert os.path.isdir(WEIGHT_DIR), f"权重目录不存在：{WEIGHT_DIR}"
meta_files = [f for f in os.listdir(WEIGHT_DIR) if f.endswith(".meta")]
assert meta_files, "没找到 .meta"
META_NAME = meta_files[0]
CKPT_BASENAME = META_NAME[:-5]

# --------- 关键：用同一个默认图建所有节点 ----------
tf.reset_default_graph()
with tf.Graph().as_default() as graph:
    # 导入骨干图
    saver_backbone = tf.train.import_meta_graph(os.path.join(WEIGHT_DIR, META_NAME), clear_devices=True)
    g = tf.get_default_graph()  # 就是 graph

    # 主输入 + 尺寸
    input_t = None
    four_d_ph = []
    for op in g.get_operations():
        for t in op.values():
            if t.op.type == "Placeholder" and t.dtype.name.startswith("float"):
                if t.shape.ndims == 4:
                    four_d_ph.append(t)
                    if input_t is None:
                        input_t = t
    assert input_t is not None, "找不到 4D float 输入占位符"
    H = to_int(input_t.shape.as_list()[1], 256)
    W = to_int(input_t.shape.as_list()[2], 256)
    C = to_int(input_t.shape.as_list()[3], 1)
    print("[INFO] using input:", input_t.name, (H, W, C))

    # 其他 4D 输入统一喂零
    extra_inputs = [t for t in four_d_ph if t != input_t]
    if extra_inputs:
        print("[INFO] extra placeholders detected:")
        for t in extra_inputs:
            print("   -", t.name, t.shape.as_list())

    def extra_zero_feeds(batch_size):
        feeds = {}
        for t in extra_inputs:
            sh = t.shape.as_list()
            h = to_int(sh[1], 256); w = to_int(sh[2], 256); c = to_int(sh[3], 3)
            feeds[t] = np.zeros((batch_size, h, w, c), dtype=np.float32)
        return feeds

    # 学习开关
    learn_ph = None
    for nm in ["batch_normalization_1/keras_learning_phase:0", "keras_learning_phase:0"]:
        try:
            learn_ph = g.get_tensor_by_name(nm); break
        except: pass
    if learn_ph is not None:
        print("[INFO] learning phase placeholder:", learn_ph.name)

    # 特征层
    features = None
    if FEATURE_TENSOR_OVERRIDE:
        try:
            features = g.get_tensor_by_name(FEATURE_TENSOR_OVERRIDE)
            print("[INFO] using features (override):", features.name)
        except:
            print("[WARN] override 不存在：", FEATURE_TENSOR_OVERRIDE)
    if features is None:
        for c in ["global_average_pooling2d/Mean:0", "global_average_pooling2d_1/Mean:0"]:
            try:
                features = g.get_tensor_by_name(c); print("[INFO] using features (GAP):", features.name); break
            except: pass
    if features is None:
        flat_like = list_flatten_like_tensors(g)
        if flat_like:
            features = g.get_tensor_by_name(flat_like[-1]); print("[INFO] using features (scan 2D):", features.name)
    if features is None:
        try:
            prev = g.get_tensor_by_name("norm_dense_1/MatMul:0").op.inputs[0]
            features = prev; print("[INFO] using features (prev of logits):", features.name)
        except:
            raise RuntimeError("找不到合适的特征层；请改 FEATURE_TENSOR_OVERRIDE。")

    # 新三分类头（纯 TF v1）
    with tf.variable_scope("new_head"):
        feat_dim = features.shape.as_list()[-1]
        if feat_dim is None:
            print("[ERR] 特征维度未知。候选：")
            for n in list_flatten_like_tensors(g)[:30]: print("  ", n)
            raise ValueError("请把 FEATURE_TENSOR_OVERRIDE 设为上面的某个 2D 张量名。")
        W_new = tf.get_variable("W", shape=[feat_dim, 3], initializer=tf.glorot_uniform_initializer())
        B_new = tf.get_variable("b", shape=[3], initializer=tf.zeros_initializer())
        logits = tf.matmul(features, W_new) + B_new
        probs  = tf.nn.softmax(logits, name="softmax")

    # 损失与指标
    y_true = tf.placeholder(tf.int64, shape=[None], name="y")
    loss = tf.reduce_mean(tf.nn.sparse_softmax_cross_entropy_with_logits(labels=y_true, logits=logits))
    preds = tf.argmax(probs, axis=1, output_type=tf.int64)
    acc   = tf.reduce_mean(tf.cast(tf.equal(preds, y_true), tf.float32))

    # 优化器与训练 op（在同一默认图里建）
    if FREEZE_BACKBONE:
        train_vars = tf.get_collection(tf.GraphKeys.TRAINABLE_VARIABLES, scope="new_head")
    else:
        all_vars = tf.trainable_variables()
        train_vars = [v for v in all_vars if not v.name.startswith("norm_dense_")]
    optimizer = tf.train.AdamOptimizer(LR, name="AdamOpt")
    train_op  = optimizer.minimize(loss, var_list=train_vars)

    # Saver（全图）
    saver_full = tf.train.Saver()

    # Session 也用同一个 graph
    with tf.Session(graph=graph) as sess:
        # 先恢复骨干
        saver_backbone.restore(sess, os.path.join(WEIGHT_DIR, CKPT_BASENAME))
        # init 新头
        sess.run(tf.variables_initializer(tf.global_variables(scope="new_head")))
        # 再把未初始化的（Adam 的 beta1/2_power、slot 等）补 init
        gvars = tf.global_variables()
        is_init = sess.run([tf.is_variable_initialized(v) for v in gvars])
        uninit = [v for v, ok in zip(gvars, is_init) if not ok]
        if uninit: sess.run(tf.variables_initializer(uninit))

        # 训练/验证
        def run_epoch(lst, train_mode=True):
            losses, accs = [], []
            for files, labs in batch_iter(lst, BATCH_SIZE, shuffle=train_mode):
                X = load_images(files, H, W, C)
                feed = {input_t: X, y_true: labs}
                if learn_ph is not None: feed[learn_ph] = 1 if train_mode else 0
                feed.update(extra_zero_feeds(X.shape[0]))
                if train_mode:
                    l, a, _ = sess.run([loss, acc, train_op], feed_dict=feed)
                else:
                    l, a = sess.run([loss, acc], feed_dict=feed)
                losses.append(l); accs.append(a)
            return float(np.mean(losses)), float(np.mean(accs)) if accs else 0.0

        best_val = 0.0
        os.makedirs("ckpt_3class", exist_ok=True)
        for ep in range(1, EPOCHS+1):
            tr_l, tr_a = run_epoch(train_list, True)
            va_l, va_a = run_epoch(val_list,   False)
            print(f"[E{ep:02d}] train loss {tr_l:.4f} acc {tr_a:.4f} | val loss {va_l:.4f} acc {va_a:.4f}")
            if va_a > best_val:
                best_val = va_a
                saver_full.save(sess, "ckpt_3class/covidnet_cxr3_3class")
                print(f"    ↳ saved (epoch {ep})")

        # 评估
        yT, yP = [], []
        bs = 32
        for i in range(0, len(val_list), bs):
            batch = val_list[i:i+bs]
            X = load_images([p for p,_ in batch], H, W, C)
            labs = np.array([lab for _,lab in batch], np.int64)
            feed = {input_t: X}
            if learn_ph is not None: feed[learn_ph] = 0
            feed.update(extra_zero_feeds(X.shape[0]))
            pr = sess.run(probs, feed_dict=feed)
            yP.extend(np.argmax(pr, axis=1).tolist())
            yT.extend(labs.tolist())

        yT, yP = np.array(yT), np.array(yP)
        cm = np.zeros((3,3), dtype=np.int32)
        for t, p in zip(yT, yP): cm[t,p]+=1
        per_class_acc = cm.diagonal()/np.maximum(1, cm.sum(axis=1))
        print("Confusion Matrix (rows=true, cols=pred):\n", cm)
        print("Per-class acc [normal, pneumonia, COVID-19]:", np.round(per_class_acc, 4))
        print("Overall acc:", np.round((yT==yP).mean(), 4))
        print("[DONE] 全流程结束，权重在 ckpt_3class/")
```
